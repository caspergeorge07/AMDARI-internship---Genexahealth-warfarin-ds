```python
# Core data handling libraries
import pandas as pd
import numpy as np

# Visualisation libraries
import matplotlib.pyplot as plt
import seaborn as sns

# Path handling (safer than hard-coding paths)
from pathlib import Path

# Display settings so we can see more columns in tables
pd.set_option("display.max_columns", 200)

# Set a clean visual style for plots
sns.set_style("whitegrid")
```

```python
# Define the path to the merged dataset
# ../ means “go up one folder” from notebooks/ to the project root
DATA_PATH = Path("../data/processed/merged_patient_dataset.csv")

# Read the CSV into a pandas DataFrame
df = pd.read_csv(DATA_PATH)

# Show the shape (rows, columns) to confirm it loaded correctly
df.shape
```

```python
df.head() # Display the first 5 rows

```

```python
df.info() # Show column names, data types, and missing values
```

```python
# Calculate percentage of missing values per column
missing_pct = (
    df.isna()                 # True/False for missing values
      .mean()                 # Convert to proportion
      .sort_values(ascending=False)  # Most missing at the top
      .to_frame("missing_pct")
)

# Display the top 20 columns with the most missing data
missing_pct.head(20)
```

```python
# Number of unique patient IDs
unique_patients = df["patient_id"].nunique()

# Total number of rows
total_rows = len(df)

unique_patients, total_rows
```

```python
# Count exact duplicate rows
df.duplicated().sum()
```

```python
# Make a copy so we never overwrite the original merged dataset
df_clean = df.copy()

# Remove exact duplicate rows
df_clean = df_clean.drop_duplicates()

# Replace common placeholder strings with proper missing values
df_clean = df_clean.replace(
    ["", "NA", "N/A", "null", "None"],
    np.nan
)

# Confirm shape after cleaning
df_clean.shape
```

```python
# Define output folder for curated (cleaned) data
OUT_PATH = Path("../data/curated")

# Create the folder if it doesn't exist
OUT_PATH.mkdir(exist_ok=True)

# Save the cleaned dataset
df_clean.to_csv(
    OUT_PATH / "merged_patient_dataset_curated.csv",
    index=False
)

print("✅ Curated dataset saved successfully")
```

### Initial Observations
- Dataset contains ~50,000 patients
- Missingness is highest in genomics-related features
- Patient ID integrity preserved during merge
- No major duplication issues detected
