```python
# Core data handling libraries
import pandas as pd
import numpy as np

# Visualisation libraries
import matplotlib.pyplot as plt
import seaborn as sns

# Path handling (safer than hard-coding paths)
from pathlib import Path

# Display settings so we can see more columns in tables
pd.set_option("display.max_columns", 200)

# Set a clean visual style for plots
sns.set_style("whitegrid")
```

```python
# Define the path to the merged dataset
# ../ means “go up one folder” from notebooks/ to the project root
DATA_PATH = Path("../data/processed/merged_patient_dataset.csv")

# Read the CSV into a pandas DataFrame
df = pd.read_csv(DATA_PATH)

# Show the shape (rows, columns) to confirm it loaded correctly
df.shape
```

```python
df.head() # Display the first 5 rows

```

```python
df.info() # Show column names, data types, and missing values
```

```python
# Calculate percentage of missing values per column
missing_pct = (
    df.isna()                 # True/False for missing values
      .mean()                 # Convert to proportion
      .sort_values(ascending=False)  # Most missing at the top
      .to_frame("missing_pct")
)

# Display the top 20 columns with the most missing data
missing_pct.head(20)
```

```python
# Number of unique patient IDs
unique_patients = df["patient_id"].nunique()

# Total number of rows
total_rows = len(df)

unique_patients, total_rows
```

```python
# Count exact duplicate rows
df.duplicated().sum()
```

```python
# Make a copy so we never overwrite the original merged dataset
df_clean = df.copy()

# Remove exact duplicate rows
df_clean = df_clean.drop_duplicates()

# Replace common placeholder strings with proper missing values
df_clean = df_clean.replace(
    ["", "NA", "N/A", "null", "None"],
    np.nan
)

# Confirm shape after cleaning
df_clean.shape
```

```python
# Define output folder for curated (cleaned) data
OUT_PATH = Path("../data/curated")

# Create the folder if it doesn't exist
OUT_PATH.mkdir(exist_ok=True)

# Save the cleaned dataset
df_clean.to_csv(
    OUT_PATH / "merged_patient_dataset_curated.csv",
    index=False
)

print("✅ Curated dataset saved successfully")
```

### Initial Observations
- Dataset contains ~50,000 patients
- Missingness is highest in genomics-related features
- Patient ID integrity preserved during merge
- No major duplication issues detected

```python
# --------------------------------------------
# Handle missing values thoughtfully (not blindly)
# --------------------------------------------

df_prepared = df_clean.copy()

# 1️⃣ Adverse_Event
# Missing likely means "no adverse event recorded"
df_prepared["Adverse_Event_Flag"] = df_prepared["Adverse_Event"].notna().astype(int)

# Keep original column for transparency
# (do NOT overwrite it yet)

# 2️⃣ Alcohol_Intake
# Replace missing values with a clear category
df_prepared["Alcohol_Intake"] = df_prepared["Alcohol_Intake"].fillna("Unknown")

# Quick check
df_prepared[["Adverse_Event", "Adverse_Event_Flag", "Alcohol_Intake"]].head()
```

```python
missing_pct = (
    df_prepared.isna()
      .mean()
      .sort_values(ascending=False)
      .to_frame("missing_pct")
      .head(10)
)

# Display the top 20 columns with the most missing data
missing_pct.head(20)
```

```python
OUT_PATH = Path("../data/curated")

df_prepared.to_csv(
    OUT_PATH / "merged_patient_dataset_prepared.csv",
    index=False
)

print("✅ Prepared dataset saved (ready for modelling)")
```

### Missing Data Handling Summary

- The dataset is largely complete across clinical and genomic variables.
- Significant missingness exists in `Adverse_Event` (~83%), which is expected as adverse events are only recorded when they occur.
- A binary indicator `Adverse_Event_Flag` was created to capture the presence or absence of a recorded adverse event.
- Missing values in `Alcohol_Intake` were replaced with `"Unknown"` to preserve row completeness while maintaining interpretability.
- No rows were dropped due to missing data.

## 3) Genomics EDA (Clinically Critical)
These genetic markers are core drivers of dose variability:
- `CYP2C9`
- `VKORC1`
- `CYP4F2`

We will:
1) Check genotype frequencies (class balance / rare categories)
2) Check dose differences across genotypes (early signal check)
3) Note any rare levels that may need grouping later

```python
# ----------------------------
# Genotype frequency checks
# ----------------------------
genetic_cols = ["CYP2C9", "VKORC1", "CYP4F2"]

for gene in genetic_cols:
    print("\n" + "="*60)
    print(f"Gene: {gene}")
    
    # Count how many patients fall into each genotype category
    counts = df[gene].value_counts(dropna=False)
    
    # Convert to percentage for easier interpretation
    perc = (counts / len(df) * 100).round(2)
    
    # Combine into a single table
    freq_table = pd.DataFrame({"count": counts, "percent": perc})
    
    display(freq_table)
```

1️⃣ Interpretation of Your Results (Plain English, Business-Aligned)
A. Genotype frequencies — are they usable?

Yes. Very usable.

🔬 CYP2C9

*1/*1 ≈ 51% → most patients are “normal metabolizers”

*1/*2, *1/*3 ≈ 18% each → common reduced-function variants

*2/*2, *2/*3, *3/*3 → smaller but not rare enough to discard

👉 Conclusion:
CYP2C9 is well distributed. No category is so rare that it must be dropped.
This is ideal for modelling.

🔬 VKORC1

G/G ≈ 49%

A/G ≈ 36.6%

A/A ≈ 14.4%

👉 Conclusion:
Very clean genotype balance. This is exactly what you expect in warfarin-related datasets and strongly supports dose modelling.

🔬 CYP4F2

C/C ≈ 60%

C/T ≈ 30%

T/T ≈ 10%

👉 Conclusion:
Again, healthy distribution. No red flags.

```python
# ----------------------------
# Dose vs genotype (signal check)
# Boxplots help us see median shifts and spread differences.
# ----------------------------

plt.figure(figsize=(14, 4))

for i, gene in enumerate(genetic_cols, start=1):
    plt.subplot(1, 3, i)
    sns.boxplot(x=df[gene], y=df[target_col])
    plt.title(f"Dose vs {gene}")
    plt.xlabel(gene)
    plt.ylabel("Final Stable Dose (mg)")
    plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

B. Dose vs genotype plots — is there signal?

Yes — clear, clinically meaningful signal.

CYP2C9 boxplot

*1/*1 → higher median dose

*1/*3, *3/*3 → lower median dose

Spread decreases as metabolism decreases

👉 Interpretation:
Reduced-function alleles → reduced dose requirement
✔ This matches known pharmacogenetics

VKORC1 boxplot

G/G → highest dose

A/A → lowest dose

Clear separation between groups

👉 Interpretation:
VKORC1 shows one of the strongest dose effects
✔ This is a major business driver

CYP4F2 boxplot

Smaller effect size

Still a visible shift in median dose

👉 Interpretation:
CYP4F2 contributes incremental signal (useful in multivariate models)

### Numeric Feature Correlations
